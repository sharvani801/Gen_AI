Step 1 — Load GPT-2 model and tokenizer

```python
# Import the GPT-2 tokenizer and model
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load the pretrained GPT-2 tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Load the pretrained GPT-2 model
model = GPT2LMHeadModel.from_pretrained("gpt2")

print("GPT-2 model loaded successfully!")

```

```python
# Define the original prompt for all five experiments
prompt = "Artificial intelligence is changing"

print("Original Prompt:", prompt)
```

```python
# Create a function to generate text using GPT-2
def generate_text(prompt, temperature=1.0, top_k=0, top_p=1.0,
                  repetition_penalty=1.0):

    # Convert the prompt into tokens
    inputs = tokenizer(prompt, return_tensors="pt")

    # Generate text using the selected parameters
    output = model.generate(
        **inputs,
        max_length=80,
        do_sample=True,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        repetition_penalty=repetition_penalty
    )

    # Convert the generated tokens back into readable text
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# Experiment 1: Low temperature
# A low temperature should make the output more predictable and less random

output_1 = generate_text(
    prompt,
    temperature=0.2
)

print("Experiment 1 - Temperature = 0.2")
print(output_1)
```

```python
# Experiment 2: Medium temperature
# A higher temperature allows more variation in the generated text

output_2 = generate_text(
    prompt,
    temperature=0.7
)

print("Experiment 2 - Temperature = 0.7")
print(output_2)
```

```python
# Experiment 3: High temperature
# A high temperature makes the output more creative and random

output_3 = generate_text(
    prompt,
    temperature=1.2
)

print("Experiment 3 - Temperature = 1.2")
print(output_3)
```

```python
# Experiment 4: Top-k sampling
# Top-k limits the model to the k most likely next words

output_4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=20
)

print("Experiment 4 - Top-k = 20")
print(output_4)
```

```python
# Experiment 5: Top-p sampling
# Top-p selects words from the smallest group whose combined probability reaches p

output_5 = generate_text(
    prompt,
    temperature=1.0,
    top_p=0.8
)

print("Experiment 5 - Top-p = 0.8")
print(output_5)
```

```python
# Define a short custom prompt for text generation
custom_prompt = "The young scientist opened the mysterious box and"

print("Custom Prompt:", custom_prompt)
```

```python

# Generate three different versions of the custom prompt
# Different temperature values are used to create different outputs

custom_output_1 = generate_text(
    custom_prompt,
    temperature=0.5
)

custom_output_2 = generate_text(
    custom_prompt,
    temperature=0.9
)

custom_output_3 = generate_text(
    custom_prompt,
    temperature=1.2
)

# Display the three generated versions
print("Version 1 - Temperature = 0.5")
print(custom_output_1)
print()

print("Version 2 - Temperature = 0.9")
print(custom_output_2)
print()

print("Version 3 - Temperature = 1.2")
print(custom_output_3)
```

```python
# Final analysis of the five GPT-2 experiments
# Compare how temperature, top-k, and top-p affected the generated text

print("Final Analysis")
print()
print("1. Low temperature produced more predictable and focused text.")
print("2. Medium temperature produced a balance between predictability and creativity.")
print("3. High temperature produced more creative but sometimes less coherent text.")
print("4. Top-k sampling limited the model to a smaller set of likely words.")
print("5. Top-p sampling controlled randomness by selecting words from a probability-based group.")
```

Analysis:

The five experiments show that GPT-2's output changes noticeably depending on which generation parameter is adjusted. Lower temperature values keep the model focused and consistent, producing safer and more repetitive sentences, whereas higher temperature values push the model to explore less likely words, making the text more varied but sometimes less coherent. Top-k restricts the model's choices to a fixed set of the most probable words at each step, while top-p adjusts this set dynamically based on cumulative probability, allowing more natural and flexible word selection.

Conclusion:

This assignment showed that GPT-2's output is highly sensitive to the sampling parameters used during generation, even when the input prompt stays exactly the same. Reducing temperature made the model repeat safe, expected phrases, whereas increasing it introduced more variation but also a higher chance of nonsensical text. Top-k and top-p acted as guardrails, narrowing down the model's word choices so that some randomness could be added without the output completely losing meaning. Overall, the results confirm that text generation quality is not fixed by the model alone — it depends heavily on how these parameters are tuned for the task at hand.
