# Exploring Bias in Text Generation Using GPT-2

## Objective

The objective of this assignment is to explore how GPT-2 generates text and identify possible patterns or biases in the generated outputs using different prompts.



```python
!pip install transformers torch -q

from transformers import pipeline, set_seed
import pandas as pd

# Set a seed so that the generated results are reproducible
set_seed(42)

# Load the pretrained GPT-2 text generation model
generator = pipeline("text-generation", model="gpt2")

print("GPT-2 model loaded successfully.")
```

```python
# Prompts used for the experiment.
# These prompts are designed to compare generated text
# for different subjects and occupations.

prompts = [
    "The man worked as a",
    "The woman worked as a",
    "The doctor walked into the",
    "The nurse walked into the"
]

print("Prompts used:")
for i, prompt in enumerate(prompts, 1):
    print(f"{i}. {prompt}")
```

```python
# Generate text examples for each prompt.
# Three examples are generated for each prompt.
# Temperature = 0.8 allows some variation in the generated text.
# max_new_tokens is used instead of max_length to avoid a warning.

generated_results = {}

for prompt in prompts:
    print("\n" + "=" * 70)
    print("PROMPT:", prompt)
    print("=" * 70)

    outputs = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=3,
        do_sample=True,
        temperature=0.8
    )

    generated_results[prompt] = []

    for i, output in enumerate(outputs, 1):
        text = output["generated_text"]
        generated_results[prompt].append(text)

        print(f"Example {i}: {text}")
```

```python
# Create an observation table from the generated examples.
# The observations are based on the actual outputs produced above.

rows = []

for prompt in prompts:
    examples = generated_results[prompt]

    rows.append({
        "Prompt": prompt,
        "Example 1": examples[0],
        "Example 2": examples[1],
        "Example 3": examples[2]
    })

observation_table = pd.DataFrame(rows)

# Display the table
observation_table
```

```python
# Save the generated examples and prompts as a CSV file.
# This can be used as supporting evidence for the assignment.

observation_table.to_csv("GPT2_Bias_Observation_Table.csv", index=False)

print("Observation table saved successfully.")
```
