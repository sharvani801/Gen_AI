```python
# Import the GPT-2 tokenizer and model
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load the pretrained GPT-2 tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Load the pretrained GPT-2 model
model = GPT2LMHeadModel.from_pretrained("gpt2")

print("GPT-2 model loaded successfully!")
```

```python
# Define the original prompt for all five experiments
prompt = "Artificial intelligence is changing"

print("Original Prompt:", prompt)
```

```python
# Create a function to generate text using GPT-2
def generate_text(prompt, temperature=1.0, top_k=0, top_p=1.0,
                  repetition_penalty=1.0):

    # Convert the prompt into tokens
    inputs = tokenizer(prompt, return_tensors="pt")

    # Generate text using the selected parameters
    output = model.generate(
        **inputs,
        max_length=80,
        do_sample=True,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        repetition_penalty=repetition_penalty
    )

    # Convert the generated tokens back into readable text
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# Experiment 1: Low temperature
# A low temperature should make the output more predictable and less random

output_1 = generate_text(
    prompt,
    temperature=0.2
)

print("Experiment 1 - Temperature = 0.2")
print(output_1)
```

```python
# Experiment 2: Medium temperature
# A higher temperature allows more variation in the generated text

output_2 = generate_text(
    prompt,
    temperature=0.7
)

print("Experiment 2 - Temperature = 0.7")
print(output_2)
```

```python
# Experiment 3: High temperature
# A high temperature makes the output more creative and random

output_3 = generate_text(
    prompt,
    temperature=1.2
)

print("Experiment 3 - Temperature = 1.2")
print(output_3)
```

```python
# Experiment 4: Top-k sampling
# Top-k limits the model to the k most likely next words

output_4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=20
)

print("Experiment 4 - Top-k = 20")
print(output_4)
```

```python
# Experiment 5: Top-p sampling
# Top-p selects words from the smallest group whose combined probability reaches p

output_5 = generate_text(
    prompt,
    temperature=1.0,
    top_p=0.8
)

print("Experiment 5 - Top-p = 0.8")
print(output_5)
```
